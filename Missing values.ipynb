```python
# Missing Values in Pandas
# 1. What are Missing Values?

# Missing values are data points that are not available or have not been recorded in a dataset.

# For example, consider a student dataset:

# Name	Age	Marks	City
# Rahul	20	85	Delhi
# Priya	21	NaN	Mumbai
# Amit	NaN	78	Delhi
# Neha	22	90	NaN

# Here, NaN represents a missing value.

# In Pandas, missing data can commonly be represented by:

# NaN — Not a Number
# None — Python's null value
# NaT — Not a Time, used for missing datetime values
# pd.NA — Pandas' newer missing-value indicator

# 2. Why do Missing Values Occur?

# Missing values can occur for several reasons:

# Data was not entered

# Age = NaN
# Data was unavailable
# Data collection errors
# Different datasets were combined, resulting in unmatched records.
# A value does not apply to a particular observation.

# Missing values can cause problems during data analysis and machine learning, so they generally need to be handled appropriately.
```

```python
#3. Creating a DataFrame with Missing Values
import pandas as pd
import numpy as np

data = {
    'Name': ['Rahul', 'Priya', 'Amit', 'Neha', 'Ravi'],
    'Age': [20, 21, np.nan, 22, 23],
    'Marks': [85, np.nan, 78, 90, np.nan],
    'City': ['Delhi', 'Mumbai', 'Delhi', np.nan, 'Noida']
}

df = pd.DataFrame(data)

print(df)
```

```python
# 4. How to Identify Missing Values in Pandas?
# A. isnull()

# The isnull() function checks whether values are missing.
print(df.isnull())
```

```python
# B. isna()

# isna() does exactly the same thing as isnull().

print(df.isna())
```

```python
# C. Count Missing Values in Each Column

# A very useful technique is:

print(df.isnull().sum())
```

```python
#D. Find the Total Number of Missing Values

print(df.isnull().sum().sum())
```

```python
#E. Find Rows Containing Missing Values
print(df[df.isnull().any(axis=1)])
```

```python
#F. Check Whether Any Missing Value Exists
print(df.isnull().any().any())
```

```python
#6. Deleting Missing Values
df_new = df.dropna()

print(df_new)
```

```python
#B. Delete Rows Only If All Values Are Missing
df.dropna(how='all')
```

```python
#C. Delete Columns Containing Missing Values
df.dropna(axis=1)
```

```python
#7. Imputing Missing Values
df['Marks'] = df['Marks'].fillna(0)
print(df['Marks'])
```

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print(df['Marks'])
```
