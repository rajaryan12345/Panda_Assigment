```python
import pandas as pd

# Load the Netflix CSV file
df = pd.read_csv("netflix.csv")

# Display the dataframe
df
```

```python
# a) Print first 5 and last 5 rows

print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())
```

```python
# b) Check number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# c) Print all column names
for col in df.columns:
    print(col)
```

```python
# d) Descriptive statistics of all variables
df.describe(include='all').T
```

```python
# e) Number of unique values for each column
df.nunique()
```

```python
# f) Percentage of missing values for each column
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

print(missing_percentage)
```

```python
# g) Delete rows where Director is missing
print("Missing Director values:", df['director'].isnull().sum())

# Then delete those rows:
df = df.dropna(subset=['director'])

# Check the new size:
print("New shape:", df.shape)
```

```python
# h) Print records where country contains Germany
germany_records = df[
    df['country'].str.contains('Germany', case=False, na=False)
]

display(germany_records)
```

```python
# i) Split Duration into numeric value and string
df[['Duration_value', 'Duration_unit']] = df['duration'].str.split(' ', n=1, expand=True)

df['Duration_value'] = pd.to_numeric(
    df['Duration_value'],
    errors='coerce'
)

display(df[['duration', 'Duration_value', 'Duration_unit']].head(10))
```

```python
# j) Split Date added into date, month and year
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
df['Date'] = df['date_added'].dt.day.astype('Int64')
df['Month'] = df['date_added'].dt.month_name()
df['Year'] = df['date_added'].dt.year.astype('Int64')

display(
    df[['date_added', 'Date', 'Month', 'Year']].head(10)
)
```

```python
# k) Number of TV Shows/Movies released in each year
release_count = pd.crosstab(
    df['release_year'],
    df['type']
)

print(release_count)

release_count['Total'] = release_count.sum(axis=1)

display(release_count)

yearly_count = df['release_year'].value_counts().sort_index()

print(yearly_count)
```

```python
# l) Rename title to movie_title
df = df.rename(columns={'title': 'movie_title'})

print(df.columns)
```

```python
# m) Split listed_in into Genre1, Genre2, Genre3
df[['Genre1', 'Genre2', 'Genre3']] = df['listed_in'].str.split(
    ',',
    n=2,
    expand=True
)

df['Genre1'] = df['Genre1'].str.strip()
df['Genre2'] = df['Genre2'].str.strip()
df['Genre3'] = df['Genre3'].str.strip()

display(
    df[['listed_in', 'Genre1', 'Genre2', 'Genre3']].head(10)
)
```
