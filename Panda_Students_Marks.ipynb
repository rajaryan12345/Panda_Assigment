```python
# 1. Import libraries and CSV files
import pandas as pd
import numpy as np

students = pd.read_csv("Student.csv")
marks = pd.read_csv("Mark.csv")

print("Students DataFrame:")
display(students.head())

print("Marks DataFrame:")
display(marks.head())
```

```python
# 2. Check column names

print("Students columns:")
print(students.columns.tolist())

print("\nMarks columns:")
print(marks.columns.tolist())
```

```python
# (a) Combine both dataframes
df = pd.merge(students, marks, on="Student_id", how="outer")

print("Combined DataFrame:")
display(df)
```

```python
#(b) Maximum and minimum marks gender-wise
gender_wise_marks = df.groupby("Gender")["Mark"].agg(
    Maximum_Marks="max",
    Minimum_Marks="min"
)

print(gender_wise_marks)
```

```python
# (c) Students who scored more than class average
average_marks = df["Mark"].mean()

print("Class Average Marks:", average_marks)

above_average = df[df["Mark"] > average_marks][["Student_id", "Mark"]]

print("\nStudents scoring more than class average:")
display(above_average)
```

```python
# (d) Males who are employed
male_employed = df[
    (df["Gender"] == "Male") &
    (df["Employed"] == "yes")
]

display(male_employed)
```

```python
#(e) Create IQ_level

#The condition needs to be interpreted as:

#Marks > 80 → Intelligent
#Marks 50–80 → Mediocre
#Marks < 50 → Weak

def assign_iq_level(marks):
    if marks > 80:
        return "Intelligent"
    elif marks >= 50:
        return "Mediocre"
    else:
        return "Weak"

df["IQ_level"] = df["Mark"].apply(assign_iq_level)

display(df)
```

```python
#(f) Count males and females from each city
city_gender_count = (
    df.groupby(["City", "Gender"])
      .size()
      .reset_index(name="Count")
)

display(city_gender_count)
```

```python
#(g) Top 5 male scorers
top_5_males = (
    df[df["Gender"] == "Male"]
    .sort_values(by="Mark", ascending=False)
    .head(5)
)

display(top_5_males)
```

```python
# (h) Replace Male/Female with M/F and export
df_export = df.copy()

df_export["Gender"] = df_export["Gender"].replace({
    "Male": "M",
    "Female": "F"
})

display(df_export)

df_export.to_csv(r"C:\Users\pc\Data_Science\Assignment_Python\Panda\test.csv", index=False)
```

```python
# (i) Check duplicate Student_ID
print("========== (i) Duplicate Student_ID ==========")

duplicate_ids = df.loc[
    df["Student_id"].duplicated(keep=False),
    "Student_id"
].unique()

if len(duplicate_ids) == 0:
    print("No duplicate Student_ID found.")
else:
    print("Duplicate Student_IDs:", duplicate_ids)
```

```python
# (j) Integer/Float variables
numeric_df = df.select_dtypes(include=["int64", "int32", "float64", "float32"])

print("Integer/Float variables:")
display(numeric_df)
```

```python
# (k) IDs present in Students but not Marks
missing_in_marks = students[
    ~students["Student_id"].isin(marks["Student_id"])
]

print("Students present in Students table but not in Marks table:")
display(missing_in_marks[["Student_id"]])
```
